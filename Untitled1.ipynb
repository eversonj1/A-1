Objective

The purpose of this project is to implement a cosine similarity-based function to compare text documents using term frequencies. This project demonstrates how text can be converted into numerical vectors and compared to measure similarity.

````python
import re
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
````

````python
text1 = "The quick brown fox jumps over the lazy dog."
text2 = "A quick brown fox leaps over a lazy dog."
text3 = "The beach is sunny and perfect for swimming."
text4 = "Quantum mechanics studies matter and energy at very small scales."

texts = [text1, text2, text3, text4]

for i, text in enumerate(texts, 1):
    print(f"Text {i}: {text}")
````

Text Samples

This project uses four short text samples. Text 1 and Text 2 are similar because they describe nearly the same idea using slightly different wording. Text 3 and Text 4 are different in topic and vocabulary, which should lead to lower similarity scores.

````python
def preprocess_text(text):
    text = text.lower()
    text = re.sub(r'[^a-z0-9\s]', '', text)
    return text
````

````python
cleaned_texts = [preprocess_text(text) for text in texts]

for i, text in enumerate(cleaned_texts, 1):
    print(f"Cleaned Text {i}: {text}")
````


Preprocessing

Before comparing the texts, the text was preprocessed by converting all characters to lowercase and removing punctuation. This step helps reduce noise and makes the comparison more consistent.
```
# This is formatted as code
```



````python
vectorizer = CountVectorizer()
term_matrix = vectorizer.fit_transform(cleaned_texts)

feature_names = vectorizer.get_feature_names_out()
term_df = pd.DataFrame(term_matrix.toarray(), columns=feature_names, index=[f"Text {i}" for i in range(1, 5)])

term_df
````

Term Frequency Representation

The texts were converted into vectors using term frequencies. Each unique word became a feature, and the value in each feature shows how often that word appears in the text.

````python
def calculate_cosine_similarity(text_a, text_b):
    cleaned = [preprocess_text(text_a), preprocess_text(text_b)]
    vectorizer = CountVectorizer()
    matrix = vectorizer.fit_transform(cleaned)
    similarity = cosine_similarity(matrix[0:1], matrix[1:2])[0][0]
    return similarity
````

````python
pairs = [
    ("Text 1 vs Text 2", text1, text2),
    ("Text 1 vs Text 3", text1, text3),
    ("Text 1 vs Text 4", text1, text4),
    ("Text 3 vs Text 4", text3, text4)
]

for label, a, b in pairs:
    score = calculate_cosine_similarity(a, b)
    print(f"{label}: {score:.4f}")
````

````python
similarity_matrix = cosine_similarity(term_matrix)

similarity_df = pd.DataFrame(
    similarity_matrix,
    index=[f"Text {i}" for i in range(1, 5)],
    columns=[f"Text {i}" for i in range(1, 5)]
)

similarity_df
````

Cosine Similarity Results

Cosine similarity measures how close two text vectors are by comparing the angle between them. A higher score means the texts are more similar, while a lower score means they are less related.

````python
print("Interpretation:")
print("- Text 1 and Text 2 should have the highest similarity because they discuss the same idea.")
print("- Text 1 and Text 3 should have a lower similarity because they describe different topics.")
print("- Text 1 and Text 4 should have a very low similarity because they are unrelated.")
print("- Text 3 and Text 4 should also show low similarity because they focus on different subjects.")
````

Findings

The cosine similarity scores showed that the most similar pair of texts was Text 1 and Text 2 because they used many of the same words and expressed nearly the same meaning. The other text pairs had lower scores because they discussed different topics. This demonstrates how cosine similarity can distinguish between similar and dissimilar texts using term frequency vectors.

Conclusion
