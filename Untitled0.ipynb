```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.util import ngrams

nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
text = """
Natural language processing helps computers understand human language.
N-grams are useful for text prediction and language modeling.
"""

print(text)
```

```python
text = """
Natural language processing helps computers understand human language.
N-grams are useful for text prediction and language modeling.
"""

print(text)
```

The sample text used in this project contains two short sentences about natural language processing and N-grams. This text is simple, but it is enough to demonstrate how N-grams are created for multiple values of N.

```python
sentences = sent_tokenize(text)

print("Tokenized Sentences:")
for i, sentence in enumerate(sentences, 1):
    print(f"Sentence {i}: {sentence}")
```

```python
word_tokens = word_tokenize(text)

print("Word Tokens:")
print(word_tokens)
```

Tokenization is the process of splitting text into smaller parts. In this project, the text was tokenized into both sentences and words. Sentence tokenization separates the text into individual sentences, while word tokenization separates it into individual words and punctuation marks.

```python
unigrams = list(ngrams(word_tokens, 1))

print("Unigrams:")
for gram in unigrams:
    print(gram)
```

```python
bigrams = list(ngrams(word_tokens, 2))

print("Bigrams:")
for gram in bigrams:
    print(gram)
```

```python
trigrams = list(ngrams(word_tokens, 3))

print("Trigrams:")
for gram in trigrams:
    print(gram)
```

A unigram contains one token at a time. A bigram contains two consecutive tokens. A trigram contains three consecutive tokens. As the value of N increases, the model captures more local context, but it also becomes more specific and less flexible.

```python
print(f"Number of unigrams: {len(unigrams)}")
print(f"Number of bigrams: {len(bigrams)}")
print(f"Number of trigrams: {len(trigrams)}")
```

The unigram model produced individual tokens, which is useful for understanding word frequency. The bigram model captured short two-word relationships, while the trigram model captured slightly longer word patterns. This shows how increasing N gives more context, but also reduces the total number of N-grams available from a short text sample.

This project demonstrated how N-grams can be created from tokenized text using NLTK. By generating unigrams, bigrams, and trigrams, it becomes easier to see how text can be represented for NLP tasks such as text prediction, autocomplete, and language modeling.
