```python
import os
```

```python
def extract_file_details(filename):
    """
    Extracts the base filename and extension from a file path or filename.

    Parameters:
        filename (str): A file path or filename.

    Returns:
        tuple: (name_without_extension, extension)
    """


    base_name = os.path.basename(filename)


    if base_name.startswith(".") and base_name.count(".") == 1:
        return base_name, ""


    name, extension = os.path.splitext(base_name)


    extension = extension[1:] if extension.startswith(".") else extension

    return name, extension
```

```python
test_files = [
    "meeting_notes_2023.pdf",
    "report.docx",
    "archive.tar.gz",
    "notes",
    "/home/user/documents/project_plan.txt",
    "my.file.name.csv",
    ".gitignore",
    "data_backup_2024.final.xlsx"
]

for file in test_files:
    name, ext = extract_file_details(file)
    print(f"Input: {file}")
    print(f"Filename without extension: {name}")
    print(f"Extension: {ext}")
    print("-" * 50)
```

```python
edge_cases = [
    "filename.with.many.dots.pdf",
    "no_extension_file",
    ".env",
    "folder/subfolder/image.jpeg",
    "backup.tar.gz"
]

for file in edge_cases:
    name, ext = extract_file_details(file)
    print(f"Input: {file} -> Name: {name}, Extension: {ext}")
```
