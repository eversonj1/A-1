The purpose of this project is to apply text preprocessing techniques to user comments. These techniques include tokenization, stop word removal, normalization, and lemmatization. The processed text can then be used for later NLP tasks such as sentiment analysis or topic modeling.

```python
comments = [
    "I absolutely loved this article! It was very helpful and easy to understand.",
    "Terrible recommendation system... it keeps showing me stuff I do not care about.",
    "The content is okay, but there are too many ads on the website!",
    "Amazing work! I’ve been reading more because of these recommendations.",
    "Not bad, but the videos load slowly and sometimes freeze.",
    "I can't believe how useful this content is. Great job!",
    "Worst experience ever. Nothing matches my interests.",
    "Pretty good overall, but I wish there were more sports-related articles."
]

for i, comment in enumerate(comments, 1):
    print(f"Comment {i}: {comment}")
```

```python
import re
import spacy
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')

nlp = spacy.load("en_core_web_sm")
stop_words = set(stopwords.words("english"))

print("Setup complete.")
```

These are the raw user comments before preprocessing. They include punctuation, capitalization, contractions, and other text variations that create noise for NLP tasks.

```python
tokenized_comments = []

for comment in comments:
    doc = nlp(comment)
    tokens = [token.text for token in doc]
    tokenized_comments.append(tokens)

for i, tokens in enumerate(tokenized_comments, 1):
    print(f"Comment {i} tokens:")
    print(tokens)
    print()
```

Tokenization breaks text into smaller units, usually words or sentences. This helps prepare the text for later NLP steps.

```python
sample_comment = comments[0]
doc = nlp(sample_comment)

sentences = [sent.text for sent in doc.sents]

print("Original comment:")
print(sample_comment)
print("\nSentence tokens:")
print(sentences)
```

Normalization makes text more consistent by converting text to lowercase and removing punctuation and special characters.

```python
def normalize_text(text):
    text = text.lower()
    text = re.sub(r'[^a-z0-9\s]', '', text)
    return text

normalized_comments = [normalize_text(comment) for comment in comments]

for i, comment in enumerate(normalized_comments, 1):
    print(f"Normalized Comment {i}: {comment}")
```

Stop words are common words such as “the,” “is,” and “and” that often do not add much meaning to text analysis. Removing them helps focus on more important terms.

```python
filtered_comments = []

for comment in normalized_comments:
    doc = nlp(comment)
    filtered_tokens = [token.text for token in doc if token.text not in stop_words and token.text.strip() != ""]
    filtered_comments.append(filtered_tokens)

for i, tokens in enumerate(filtered_comments, 1):
    print(f"Comment {i} after stop word removal:")
    print(tokens)
    print()
```

Lemmatization reduces words to their base or dictionary form. This helps group similar word forms together, such as “reading” and “read.”

```python
lemmatized_comments = []

for comment in normalized_comments:
    doc = nlp(comment)
    lemmas = [token.lemma_ for token in doc if token.text not in stop_words and token.text.strip() != ""]
    lemmatized_comments.append(lemmas)

for i, lemmas in enumerate(lemmatized_comments, 1):
    print(f"Comment {i} lemmatized:")
    print(lemmas)
    print()
```

The results below show how the comments change after each preprocessing step.

```python
for i in range(len(comments)):
    print(f"Original Comment {i+1}:")
    print(comments[i])
    print("Normalized:")
    print(normalized_comments[i])
    print("After Stop Word Removal:")
    print(filtered_comments[i])
    print("After Lemmatization:")
    print(lemmatized_comments[i])
    print("-" * 60)
```

```python
def preprocess_text(text):
    text = text.lower()
    text = re.sub(r'[^a-z0-9\s]', '', text)
    doc = nlp(text)
    processed = [token.lemma_ for token in doc if token.text not in stop_words and token.text.strip() != ""]
    return processed

processed_comments = [preprocess_text(comment) for comment in comments]

print("Fully processed comments:")
for i, comment in enumerate(processed_comments, 1):
    print(f"Comment {i}: {comment}")
```

The preprocessing steps reduced noise in the comments and made the text more structured for NLP analysis. Tokenization separated text into manageable pieces. Normalization made the comments more consistent. Stop word removal reduced less meaningful words, and lemmatization simplified words to their base forms. These steps made the text cleaner and more useful for future analysis.

This project showed that preprocessing is an important step in NLP. Raw comments often include noise, punctuation, and inconsistencies that make text analysis more difficult. By applying tokenization, normalization, stop word removal, and lemmatization, the text becomes easier to analyze and more useful for downstream NLP tasks.
