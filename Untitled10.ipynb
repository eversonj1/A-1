```python
# @title
# CELL 2 — Generate synthetic Tidewater Mobility trip events

from datetime import datetime, timedelta

np.random.seed(42)
random.seed(42)

# Simulate 7 days
START = pd.Timestamp("2026-09-01 00:00:00")
DAYS = 7
TRIPS_PER_DAY = 500

zones = [
    "Downtown",
    "Harbor",
    "Airport",
    "North Beach",
    "University",
    "East Side"
]

records = []
trip_number = 1

for day in range(DAYS):
    day_start = START + pd.Timedelta(days=day)

    for _ in range(TRIPS_PER_DAY):

        # Random event time during the simulated day
        seconds_into_day = np.random.randint(0, 24 * 60 * 60)
        event_time = day_start + pd.Timedelta(seconds=int(seconds_into_day))

        zone = np.random.choice(zones)

        # -------------------------------------------------
        # Lateness distribution:
        # 92% arrive within 2 minutes
        # 7% arrive between 2 and 12 minutes
        # 1% arrive between 20 and 90 minutes
        # -------------------------------------------------
        lateness_band = np.random.choice(
            ["normal", "moderate", "extreme"],
            p=[0.92, 0.07, 0.01]
        )

        if lateness_band == "normal":
            delay_seconds = np.random.randint(0, 121)

        elif lateness_band == "moderate":
            delay_seconds = np.random.randint(121, 721)

        else:
            delay_seconds = np.random.randint(1200, 5401)

        processing_time = event_time + pd.Timedelta(
            seconds=int(delay_seconds)
        )

        record = {
            "trip_id": f"T{trip_number:06d}",
            "zone": zone,
            "event_time": event_time,
            "processing_time": processing_time,
            "lateness_seconds": int(delay_seconds),
            "lateness_band": lateness_band
        }

        # Mid-stream schema evolution:
        # surge_multiplier first appears on simulated Day 4.
        if day >= 3:
            record["surge_multiplier"] = round(
                float(np.random.choice(
                    [1.0, 1.1, 1.25, 1.5, 1.75, 2.0],
                    p=[0.35, 0.20, 0.18, 0.15, 0.08, 0.04]
                )),
                2
            )

        records.append(record)
        trip_number += 1


events = pd.DataFrame(records)

# Day 1-3 did not originally contain this field.
# Pandas represents that historical absence as NaN once
# all simulated days are combined.
events["surge_multiplier"] = events.get(
    "surge_multiplier",
    pd.Series(index=events.index, dtype=float)
)

# Sort by PROCESSING TIME to simulate arrival order.
# This deliberately causes event times to appear out of order.
events = events.sort_values(
    ["processing_time", "trip_id"]
).reset_index(drop=True)

# Save generated stream
event_file = PROJECT / "data" / "events" / "trip_events.csv"
events.to_csv(event_file, index=False)

# Validation
total = len(events)
normal = (events["lateness_band"] == "normal").sum()
moderate = (events["lateness_band"] == "moderate").sum()
extreme = (events["lateness_band"] == "extreme").sum()

day4 = START + pd.Timedelta(days=3)

before_day4 = events[
    events["event_time"] < day4
]["surge_multiplier"].notna().sum()

day4_forward = events[
    events["event_time"] >= day4
]["surge_multiplier"].notna().sum()

# Check that arrival order is not event-time order
event_times = events["event_time"]
out_of_order_transitions = (
    event_times.diff().dt.total_seconds() < 0
).sum()

print("SYNTHETIC EVENT STREAM CREATED")
print("=" * 45)
print(f"Total trips:                 {total:,}")
print(f"Normal <=2 min:              {normal:,} ({normal/total:.1%})")
print(f"Moderate 2-12 min:           {moderate:,} ({moderate/total:.1%})")
print(f"Extreme 20-90 min:           {extreme:,} ({extreme/total:.1%})")
print(f"Out-of-order transitions:    {out_of_order_transitions:,}")
print()
print("SCHEMA CHANGE")
print("-" * 45)
print(f"surge_multiplier before Day 4: {before_day4:,}")
print(f"surge_multiplier Day 4 onward: {day4_forward:,}")
print()
print("Saved:", event_file)
```

```python
# @title
# CELL 3 — Micro-batch event-time consumer

WINDOW_MINUTES = 15
WATERMARK_MINUTES = 12
BATCH_SIZE = 100

events_stream = pd.read_csv(
    PROJECT / "data" / "events" / "trip_events.csv",
    parse_dates=["event_time", "processing_time"]
)

# Already sorted by processing_time, which represents arrival order.
events_stream = events_stream.sort_values(
    ["processing_time", "trip_id"]
).reset_index(drop=True)

accepted_events = []
late_side_output = []

max_event_time_seen = None

for batch_start in range(0, len(events_stream), BATCH_SIZE):

    batch = events_stream.iloc[
        batch_start:batch_start + BATCH_SIZE
    ]

    for _, event in batch.iterrows():

        event_time = event["event_time"]

        # Watermark is based on the greatest event time observed so far.
        if max_event_time_seen is None:
            current_watermark = event_time - pd.Timedelta(
                minutes=WATERMARK_MINUTES
            )
        else:
            current_watermark = max_event_time_seen - pd.Timedelta(
                minutes=WATERMARK_MINUTES
            )

        # Event arrived beyond the watermark.
        if event_time < current_watermark:

            late_record = event.to_dict()
            late_record["watermark_at_arrival"] = current_watermark
            late_record["route"] = "SIDE_OUTPUT"

            late_side_output.append(late_record)

        else:
            accepted_record = event.to_dict()
            accepted_record["watermark_at_arrival"] = current_watermark
            accepted_record["route"] = "MAIN"

            accepted_events.append(accepted_record)

        # Advance max event time AFTER evaluating the event.
        if (
            max_event_time_seen is None
            or event_time > max_event_time_seen
        ):
            max_event_time_seen = event_time


accepted_df = pd.DataFrame(accepted_events)
late_df = pd.DataFrame(late_side_output)

# Assign 15-minute EVENT-TIME windows
accepted_df["window_start"] = (
    accepted_df["event_time"].dt.floor(f"{WINDOW_MINUTES}min")
)

accepted_df["window_end"] = (
    accepted_df["window_start"]
    + pd.Timedelta(minutes=WINDOW_MINUTES)
)

# Aggregate accepted trips by EVENT TIME
event_time_counts = (
    accepted_df
    .groupby(
        ["window_start", "window_end", "zone"],
        as_index=False
    )
    .agg(
        trip_count=("trip_id", "count")
    )
)

# Save late events for reconciliation
side_output_file = (
    PROJECT / "data" / "side_output" / "late_events.csv"
)

late_df.to_csv(side_output_file, index=False)

print("EVENT-TIME CONSUMER COMPLETE")
print("=" * 45)
print(f"Watermark delay:          {WATERMARK_MINUTES} minutes")
print(f"Window size:              {WINDOW_MINUTES} minutes")
print(f"Total events:             {len(events_stream):,}")
print(f"Main-path events:         {len(accepted_df):,}")
print(f"Side-output events:       {len(late_df):,}")
print(
    f"Side-output rate:         "
    f"{len(late_df)/len(events_stream):.2%}"
)
print(f"Event-time window rows:   {len(event_time_counts):,}")
print()
print("✓ Late events preserved instead of silently dropped")
print("✓ 15-minute event-time windows created")
print("✓ Side output saved for reconciliation")
```

```python
# @title
# CELL 4 — Compare EVENT TIME vs PROCESSING TIME aggregation

# --------------------------------------------------
# EVENT-TIME COUNTS
# --------------------------------------------------

event_compare = events_stream.copy()

event_compare["window_start"] = (
    event_compare["event_time"]
    .dt.floor(f"{WINDOW_MINUTES}min")
)

event_counts = (
    event_compare
    .groupby(["window_start", "zone"])
    .size()
    .reset_index(name="event_time_count")
)


# --------------------------------------------------
# PROCESSING-TIME COUNTS
# --------------------------------------------------

processing_compare = events_stream.copy()

processing_compare["window_start"] = (
    processing_compare["processing_time"]
    .dt.floor(f"{WINDOW_MINUTES}min")
)

processing_counts = (
    processing_compare
    .groupby(["window_start", "zone"])
    .size()
    .reset_index(name="processing_time_count")
)


# --------------------------------------------------
# SIDE-BY-SIDE COMPARISON
# --------------------------------------------------

comparison = pd.merge(
    event_counts,
    processing_counts,
    on=["window_start", "zone"],
    how="outer"
).fillna(0)

comparison["event_time_count"] = (
    comparison["event_time_count"].astype(int)
)

comparison["processing_time_count"] = (
    comparison["processing_time_count"].astype(int)
)

comparison["count_difference"] = (
    comparison["processing_time_count"]
    - comparison["event_time_count"]
)

comparison["absolute_difference"] = (
    comparison["count_difference"].abs()
)

# Find windows where processing-time gives the
# most different picture from event-time.
largest_differences = (
    comparison
    .sort_values(
        "absolute_difference",
        ascending=False
    )
    .head(15)
)

different_windows = (
    comparison["count_difference"] != 0
).sum()

print("EVENT TIME vs PROCESSING TIME")
print("=" * 65)

print(
    largest_differences[
        [
            "window_start",
            "zone",
            "event_time_count",
            "processing_time_count",
            "count_difference"
        ]
    ].to_string(index=False)
)

print()
print("=" * 65)
print(f"Total window/zone combinations: {len(comparison):,}")
print(f"Windows with different counts:  {different_windows:,}")
print(
    f"Percent affected:               "
    f"{different_windows / len(comparison):.1%}"
)
```

```python
# @title
# CELL 5 — Land event-time aggregates in Delta Lake

from deltalake import DeltaTable, write_deltalake
import pyarrow as pa

DELTA_PATH = str(PROJECT / "delta" / "zone_aggregates")

# Start with the accepted event-time aggregates
delta_results = event_time_counts.copy()

# Add useful metadata
delta_results["watermark_minutes"] = WATERMARK_MINUTES
delta_results["window_minutes"] = WINDOW_MINUTES
delta_results["table_status"] = "ORIGINAL"

# Make sure timestamps are clean
delta_results["window_start"] = pd.to_datetime(
    delta_results["window_start"]
)

delta_results["window_end"] = pd.to_datetime(
    delta_results["window_end"]
)

# Remove old table if this cell is rerun
if os.path.exists(DELTA_PATH):
    shutil.rmtree(DELTA_PATH)

# Write Version 0
write_deltalake(
    DELTA_PATH,
    pa.Table.from_pandas(
        delta_results,
        preserve_index=False
    ),
    mode="overwrite"
)

delta_table = DeltaTable(DELTA_PATH)

print("DELTA LAKE TABLE CREATED")
print("=" * 55)
print(f"Delta path:          {DELTA_PATH}")
print(f"Current version:     {delta_table.version()}")
print(f"Aggregate rows:      {len(delta_results):,}")
print(f"Total counted trips: {delta_results['trip_count'].sum():,}")
print()
print("Columns:")
for column in delta_results.columns:
    print(f"  - {column}")

print()
print("✓ Event-time aggregates stored in Delta Lake")
print("✓ Version 0 is our original good snapshot")
```

```python
# @title
# CELL 6 — Demonstrate mid-stream schema evolution

raw_delta_path = str(
    PROJECT / "delta" / "trip_events"
)

if os.path.exists(raw_delta_path):
    shutil.rmtree(raw_delta_path)

day4_start = START + pd.Timedelta(days=3)

# ---------------------------------------------
# DAYS 1–3: ORIGINAL SCHEMA
# surge_multiplier does not exist yet
# ---------------------------------------------

early_events = events_stream[
    events_stream["event_time"] < day4_start
].copy()

early_events = early_events[
    [
        "trip_id",
        "zone",
        "event_time",
        "processing_time",
        "lateness_seconds",
        "lateness_band"
    ]
]

write_deltalake(
    raw_delta_path,
    pa.Table.from_pandas(
        early_events,
        preserve_index=False
    ),
    mode="overwrite"
)

before_table = DeltaTable(raw_delta_path)

print("BEFORE DAY 4")
print("=" * 45)
print(f"Delta version: {before_table.version()}")
print(f"Rows:          {len(early_events):,}")
print()
print("Schema:")
print(before_table.schema().to_arrow())
print()


# ---------------------------------------------
# DAY 4+: NEW SCHEMA
# surge_multiplier appears
# ---------------------------------------------

later_events = events_stream[
    events_stream["event_time"] >= day4_start
].copy()

later_events = later_events[
    [
        "trip_id",
        "zone",
        "event_time",
        "processing_time",
        "lateness_seconds",
        "lateness_band",
        "surge_multiplier"
    ]
]

write_deltalake(
    raw_delta_path,
    pa.Table.from_pandas(
        later_events,
        preserve_index=False
    ),
    mode="append",
    schema_mode="merge"
)

after_table = DeltaTable(raw_delta_path)

print("AFTER DAY 4")
print("=" * 45)
print(f"Delta version: {after_table.version()}")
print(f"New rows:      {len(later_events):,}")
print()
print("Schema:")
print(after_table.schema().to_arrow())
print()

print("SCHEMA EVOLUTION RESULT")
print("=" * 45)
print("Version 0: surge_multiplier did not exist")
print("Version 1: surge_multiplier added")
print()
print("✓ Delta schema evolved when Day 4 data arrived")
print("✓ Historical Days 1–3 remain readable")
```

```python
# @title
# CELL 7 — Bad correction + Delta Lake time travel

from deltalake import DeltaTable, write_deltalake
import pyarrow as pa

# --------------------------------------------------
# 1. Capture the GOOD snapshot before corruption
# --------------------------------------------------

good_table = DeltaTable(DELTA_PATH)

good_version = good_table.version()

good_snapshot = good_table.to_pandas()

good_total = int(good_snapshot["trip_count"].sum())

print("BEFORE BAD CORRECTION")
print("=" * 55)
print(f"Good Delta version: {good_version}")
print(f"Aggregate rows:      {len(good_snapshot):,}")
print(f"Total trip count:    {good_total:,}")
print()


# --------------------------------------------------
# 2. Simulate a BAD late-data correction
#
# Pretend the correction job accidentally doubles
# one entire day's aggregate counts.
# --------------------------------------------------

bad_snapshot = good_snapshot.copy()

bad_day = pd.Timestamp("2026-09-04")

bad_day_mask = (
    bad_snapshot["window_start"].dt.normalize()
    == bad_day
)

affected_rows = int(bad_day_mask.sum())

original_bad_day_total = int(
    bad_snapshot.loc[
        bad_day_mask,
        "trip_count"
    ].sum()
)

# Deliberate production mistake
bad_snapshot.loc[
    bad_day_mask,
    "trip_count"
] = (
    bad_snapshot.loc[
        bad_day_mask,
        "trip_count"
    ] * 2
)

bad_snapshot.loc[
    bad_day_mask,
    "table_status"
] = "BAD_CORRECTION"

bad_day_after = int(
    bad_snapshot.loc[
        bad_day_mask,
        "trip_count"
    ].sum()
)

# Overwrite table, creating a NEW Delta version
write_deltalake(
    DELTA_PATH,
    pa.Table.from_pandas(
        bad_snapshot,
        preserve_index=False
    ),
    mode="overwrite"
)

corrupted_table = DeltaTable(DELTA_PATH)

corrupted_version = corrupted_table.version()

corrupted_snapshot = corrupted_table.to_pandas()

corrupted_total = int(
    corrupted_snapshot["trip_count"].sum()
)

print("BAD CORRECTION APPLIED")
print("=" * 55)
print(f"Bad Delta version:      {corrupted_version}")
print(f"Day corrupted:          {bad_day.date()}")
print(f"Rows affected:          {affected_rows:,}")
print(
    f"Day count before:       "
    f"{original_bad_day_total:,}"
)
print(
    f"Day count after:        "
    f"{bad_day_after:,}"
)
print(f"Whole table before:     {good_total:,}")
print(f"Whole table after:      {corrupted_total:,}")
print()


# --------------------------------------------------
# 3. TIME TRAVEL back to the good version
# --------------------------------------------------

historical_table = DeltaTable(
    DELTA_PATH,
    version=good_version
)

historical_snapshot = historical_table.to_pandas()

historical_total = int(
    historical_snapshot["trip_count"].sum()
)

historical_day_total = int(
    historical_snapshot.loc[
        historical_snapshot[
            "window_start"
        ].dt.normalize() == bad_day,
        "trip_count"
    ].sum()
)

print("DELTA TIME TRAVEL")
print("=" * 55)
print(
    f"Current corrupted version: "
    f"{corrupted_version}"
)
print(
    f"Time-traveled to version:  "
    f"{good_version}"
)
print()
print(
    f"Corrupted Sep 4 count:     "
    f"{bad_day_after:,}"
)
print(
    f"Historical Sep 4 count:    "
    f"{historical_day_total:,}"
)
print(
    f"Historical table total:    "
    f"{historical_total:,}"
)
print()

if historical_total == good_total:
    print(
        "✓ Time travel recovered the exact "
        "pre-correction snapshot"
    )
else:
    print("✗ Historical snapshot does not match")

print(
    "✓ Bad overwrite did not destroy "
    "the previous Delta version"
)
```

```python
# @title
# CELL 8 — Late-data health signals and safe reconciliation

# Reload side output
late_health = pd.read_csv(
    PROJECT / "data" / "side_output" / "late_events.csv",
    parse_dates=[
        "event_time",
        "processing_time",
        "watermark_at_arrival"
    ]
)

# Calculate delay in minutes
late_health["delay_minutes"] = (
    (
        late_health["processing_time"]
        - late_health["event_time"]
    ).dt.total_seconds() / 60
)

# Assign the original event-time window
late_health["window_start"] = (
    late_health["event_time"]
    .dt.floor(f"{WINDOW_MINUTES}min")
)

late_health["window_end"] = (
    late_health["window_start"]
    + pd.Timedelta(minutes=WINDOW_MINUTES)
)

# --------------------------------------------------
# Health metrics
# --------------------------------------------------

side_output_count = len(late_health)
side_output_rate = (
    side_output_count / len(events_stream)
)

avg_delay = late_health["delay_minutes"].mean()
max_delay = late_health["delay_minutes"].max()

late_by_zone = (
    late_health
    .groupby("zone")
    .size()
    .reset_index(name="late_event_count")
    .sort_values(
        "late_event_count",
        ascending=False
    )
)

print("LATE-DATA HEALTH SIGNALS")
print("=" * 55)
print(f"Total events:            {len(events_stream):,}")
print(f"Side-output events:      {side_output_count:,}")
print(f"Side-output rate:        {side_output_rate:.2%}")
print(f"Average late delay:      {avg_delay:.1f} minutes")
print(f"Maximum late delay:      {max_delay:.1f} minutes")
print()

print("LATE EVENTS BY ZONE")
print("=" * 55)
print(late_by_zone.to_string(index=False))
print()

# --------------------------------------------------
# Window completeness
# --------------------------------------------------

initial_count = len(accepted_df)
reconciled_count = initial_count + side_output_count

initial_completeness = (
    initial_count / len(events_stream)
)

reconciled_completeness = (
    reconciled_count / len(events_stream)
)

print("WINDOW COMPLETENESS")
print("=" * 55)
print(f"Initial main-path trips: {initial_count:,}")
print(f"Initial completeness:    {initial_completeness:.2%}")
print(f"Late trips preserved:    {side_output_count:,}")
print(f"After reconciliation:    {reconciled_count:,}")
print(f"Final completeness:      {reconciled_completeness:.2%}")
print()

print("✓ Side-output volume can be monitored as a health signal")
print("✓ Late events remain available for safe reconciliation")
print("✓ No late event was silently discarded")
```

```python
# @title
# CELL 9 — Package working prototype and README

src_dir = PROJECT / "src"
src_dir.mkdir(exist_ok=True)

# --------------------------------------------------
# EVENT GENERATOR
# --------------------------------------------------

generator_code = r'''import pandas as pd
import numpy as np
import random
from pathlib import Path

np.random.seed(42)
random.seed(42)

START = pd.Timestamp("2026-09-01 00:00:00")
DAYS = 7
TRIPS_PER_DAY = 500

ZONES = [
    "Downtown",
    "Harbor",
    "Airport",
    "North Beach",
    "University",
    "East Side"
]

def generate_events(output_path):
    records = []
    trip_number = 1

    for day in range(DAYS):
        day_start = START + pd.Timedelta(days=day)

        for _ in range(TRIPS_PER_DAY):
            seconds = np.random.randint(0, 86400)
            event_time = day_start + pd.Timedelta(
                seconds=int(seconds)
            )

            zone = np.random.choice(ZONES)

            band = np.random.choice(
                ["normal", "moderate", "extreme"],
                p=[0.92, 0.07, 0.01]
            )

            if band == "normal":
                delay = np.random.randint(0, 121)
            elif band == "moderate":
                delay = np.random.randint(121, 721)
            else:
                delay = np.random.randint(1200, 5401)

            processing_time = (
                event_time
                + pd.Timedelta(seconds=int(delay))
            )

            record = {
                "trip_id": f"T{trip_number:06d}",
                "zone": zone,
                "event_time": event_time,
                "processing_time": processing_time,
                "lateness_seconds": int(delay),
                "lateness_band": band
            }

            # Day 4 schema change
            if day >= 3:
                record["surge_multiplier"] = round(
                    float(np.random.choice(
                        [1.0, 1.1, 1.25, 1.5, 1.75, 2.0]
                    )),
                    2
                )

            records.append(record)
            trip_number += 1

    df = pd.DataFrame(records)
    df = df.sort_values(
        ["processing_time", "trip_id"]
    )

    Path(output_path).parent.mkdir(
        parents=True,
        exist_ok=True
    )

    df.to_csv(output_path, index=False)

    print(f"Generated {len(df):,} trip events.")
    print(f"Saved to {output_path}")

    return df


if __name__ == "__main__":
    generate_events("data/events/trip_events.csv")
'''

(src_dir / "generate_events.py").write_text(generator_code)


# --------------------------------------------------
# EVENT-TIME CONSUMER
# --------------------------------------------------

consumer_code = r'''import pandas as pd
from pathlib import Path

WINDOW_MINUTES = 15
WATERMARK_MINUTES = 12

def process_events(input_path, side_output_path):
    events = pd.read_csv(
        input_path,
        parse_dates=["event_time", "processing_time"]
    )

    events = events.sort_values(
        ["processing_time", "trip_id"]
    ).reset_index(drop=True)

    accepted = []
    late = []
    max_event_time = None

    for _, event in events.iterrows():
        event_time = event["event_time"]

        if max_event_time is None:
            watermark = (
                event_time
                - pd.Timedelta(
                    minutes=WATERMARK_MINUTES
                )
            )
        else:
            watermark = (
                max_event_time
                - pd.Timedelta(
                    minutes=WATERMARK_MINUTES
                )
            )

        record = event.to_dict()
        record["watermark_at_arrival"] = watermark

        if event_time < watermark:
            record["route"] = "SIDE_OUTPUT"
            late.append(record)
        else:
            record["route"] = "MAIN"
            accepted.append(record)

        if (
            max_event_time is None
            or event_time > max_event_time
        ):
            max_event_time = event_time

    accepted_df = pd.DataFrame(accepted)
    late_df = pd.DataFrame(late)

    accepted_df["window_start"] = (
        accepted_df["event_time"]
        .dt.floor(f"{WINDOW_MINUTES}min")
    )

    accepted_df["window_end"] = (
        accepted_df["window_start"]
        + pd.Timedelta(minutes=WINDOW_MINUTES)
    )

    aggregates = (
        accepted_df
        .groupby(
            ["window_start", "window_end", "zone"],
            as_index=False
        )
        .agg(trip_count=("trip_id", "count"))
    )

    Path(side_output_path).parent.mkdir(
        parents=True,
        exist_ok=True
    )

    late_df.to_csv(side_output_path, index=False)

    print(f"Total events:       {len(events):,}")
    print(f"Main path:          {len(accepted_df):,}")
    print(f"Side output:        {len(late_df):,}")
    print(f"Aggregate rows:     {len(aggregates):,}")

    return accepted_df, late_df, aggregates


if __name__ == "__main__":
    process_events(
        "data/events/trip_events.csv",
        "data/side_output/late_events.csv"
    )
'''

(src_dir / "event_time_consumer.py").write_text(
    consumer_code
)


# --------------------------------------------------
# REQUIREMENTS
# --------------------------------------------------

requirements = """pandas
numpy
pyarrow
deltalake
"""

(PROJECT / "requirements.txt").write_text(requirements)


# --------------------------------------------------
# README
# --------------------------------------------------

readme = """# Tidewater Mobility Streaming Prototype

## Purpose

This prototype demonstrates event-time processing for Tidewater
Mobility's surge-pricing analytics.

Driver phones can buffer trip-completion events when they lose
connectivity in harbor tunnels or cellular dead zones. Processing
events only by arrival time can therefore place trips into the wrong
15-minute pricing window.

## Architecture

Synthetic Event Generator
        |
        v
Processing-Time Arrival Stream
        |
        v
Event-Time Consumer
        |
        +---- Main Path
        |       |
        |       v
        |   15-Minute Event-Time Windows
        |       |
        |       v
        |   Delta Lake Zone Aggregates
        |
        +---- Beyond Watermark
                |
                v
            Side Output
                |
                v
        Historical Reconciliation

## Event Generator

The generator creates seven simulated days with 500 trips per day,
for 3,500 total trip-completion events.

Each event contains separate:

- trip_id
- zone
- event_time
- processing_time
- lateness_seconds
- lateness_band

The simulated lateness distribution is:

- 92% arrive within 2 minutes
- 7% arrive between 2 and 12 minutes
- 1% arrive between 20 and 90 minutes

Events are sorted by processing_time to simulate the order in which
Tidewater receives them. This creates out-of-order event times.

## Schema Evolution

Days 1-3 use the original event schema.

Beginning on simulated Day 4, a new field appears:

surge_multiplier

Delta Lake schema evolution allows the field to be added without
rebuilding the historical table.

## Event-Time Policy

The prototype uses:

- 15-minute tumbling event-time windows
- 12-minute watermark delay
- side output for events arriving beyond the watermark

Events beyond the watermark are not silently discarded. They are
written to a late-event side output for reconciliation and monitoring.

## Event Time vs Processing Time

Event time represents when the trip actually occurred.

Processing time represents when Tidewater received the event.

The prototype compares these two approaches because delayed mobile
events can cause processing-time aggregation to move trips into the
wrong pricing window.

In the deterministic test data:

- 2,487 window/zone combinations were observed
- 612 had different event-time and processing-time counts
- 24.6% of window/zone combinations were affected

## Late-Data Health Signals

The deterministic run produced:

- 3,500 total events
- 3,466 main-path events
- 34 side-output events
- 0.97% side-output rate
- 99.03% initial completeness
- 100% historical completeness after reconciliation

Side-output volume is a useful operational signal. A sudden increase
could indicate driver connectivity problems, tunnel buffering, mobile
network issues, or ingestion delays.

## Delta Lake and Time Travel

The event-time aggregates are stored in Delta Lake.

A simulated bad correction job doubles all September 4 aggregate
counts:

- Correct September 4 count: 494
- Corrupted count: 988

The overwrite creates a new Delta version rather than destroying the
previous snapshot.

A time-travel query against Delta Version 0 recovers the exact
pre-correction table, including the original September 4 count of 494.

## Business Interpretation

Symptom:
The pricing team sees undercounted or shifted trip demand in certain
15-minute windows, which can produce incorrect surge decisions.

Underlying cause:
Driver phones buffer events during connectivity loss, causing
processing_time to differ from event_time.

A separate failure mode occurs when reconciliation logic overwrites
correct historical aggregates with bad late-data corrections. Delta
Lake version history makes that failure auditable and recoverable.

## Files

- src/generate_events.py
- src/event_time_consumer.py
- data/events/trip_events.csv
- data/side_output/late_events.csv
- delta/zone_aggregates/
- delta/trip_events/
- requirements.txt
- README.md

## Run

Install dependencies:

pip install -r requirements.txt

Generate data:

python src/generate_events.py

Run consumer:

python src/event_time_consumer.py
"""

(PROJECT / "README.md").write_text(readme)


# --------------------------------------------------
# DISPLAY CREATED FILES
# --------------------------------------------------

print("SUBMISSION FILES CREATED")
print("=" * 55)

for path in [
    src_dir / "generate_events.py",
    src_dir / "event_time_consumer.py",
    PROJECT / "requirements.txt",
    PROJECT / "README.md"
]:
    print("✓", path.relative_to(PROJECT))

print()
print("✓ Generator packaged")
print("✓ Consumer packaged")
print("✓ README created")
print("✓ Requirements file created")
```
